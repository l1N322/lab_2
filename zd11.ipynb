```python
import numpy as np

A = np.array([
    [1, -2, 0],
    [0, -1, 1],
    [1, 1, 1]
])
B = np.array([0, 200, 1500])

A_det = np.linalg.det(A)
if A_det == 0:
    print("Матрица вырождена, метод Крамера неприменим.")
else:
    x_cramer = []
    for i in range(3):
        Ai = A.copy()
        Ai[:, i] = B
        Ai_det = np.linalg.det(Ai)
        x_cramer.append(Ai_det / A_det)
    x_cramer = np.array(x_cramer)
    print("Метод Крамера:", x_cramer)

if A_det == 0:
    print("Матрица вырождена, обратная матрица не существует.")
else:
    A_inv = np.linalg.inv(A)
    x_inv = np.dot(A_inv, B)
    print("Матричный метод:", x_inv)

x_solve = np.linalg.solve(A, B)
print("np.linalg.solve:", x_solve)

print("\nВсе методы дают одинаковый результат (x1, x2, x3):")
print(f"x1 = {x_solve[0]}, x2 = {x_solve[1]}, x3 = {x_solve[2]}")
```
