```python
import numpy as np

arr1 = np.random.randint(0, 51, size=(1, 5))
arr2 = np.random.randint(0, 51, size=(1, 5))

print("Первый массив:", arr1)
print("Второй массив:", arr2)

combined = np.vstack((arr1, arr2))
print("Объединённый двумерный массив:\n", combined)

split1, split2 = np.vsplit(combined, 2)
print("Первый разделённый массив:", split1)
print("Второй разделённый массив:", split2)
```
